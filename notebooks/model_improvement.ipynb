```python
from collections import Counter
import numpy as np
import ast

def normalize_skill_value(value):

    if value is None:
        return []

    if isinstance(value, float) and pd.isna(value):
        return []

    if isinstance(value, np.ndarray):
        return value.tolist()

    if isinstance(value, list):
        return value

    if isinstance(value, tuple):
        return list(value)

    if isinstance(value, str):

        value = value.strip()

        if not value:
            return []

        # Try Python-list representation
        if value.startswith("[") and value.endswith("]"):
            try:
                parsed = ast.literal_eval(value)

                if isinstance(parsed, list):
                    return parsed

            except Exception:
                pass

        # Otherwise treat as delimiter-separated
        if "|" in value:
            return [
                x.strip()
                for x in value.split("|")
                if x.strip()
            ]

        return [value]

    return []
```

```python
salary_df["skills_normalized"] = (
    salary_df["skill_list"]
    .apply(normalize_skill_value)
)
```

```python
skill_counter = Counter()

for skills in salary_df["skills_normalized"]:
    skill_counter.update(
        str(skill).strip().lower()
        for skill in skills
        if str(skill).strip()
    )

skill_counts = pd.DataFrame(
    skill_counter.items(),
    columns=["skill", "job_count"]
)

skill_counts["percentage"] = (
    skill_counts["job_count"]
    / len(salary_df)
    * 100
)

skill_counts = (
    skill_counts
    .sort_values(
        "job_count",
        ascending=False
    )
    .reset_index(drop=True)
)

display(skill_counts.head(50))
```

```python
skill_salary = []

for skill in skill_counts["skill"]:

    mask = salary_df[
        "skills_normalized"
    ].apply(
        lambda skills: skill in [
            str(x).strip().lower()
            for x in skills
        ]
    )

    subset = salary_df.loc[
        mask,
        "target_annual_salary"
    ]

    if len(subset) >= 30:

        skill_salary.append({
            "skill": skill,
            "jobs": len(subset),
            "mean_salary": subset.mean(),
            "median_salary": subset.median()
        })

skill_salary_df = (
    pd.DataFrame(skill_salary)
    .sort_values(
        "jobs",
        ascending=False
    )
)

display(skill_salary_df.head(50))
```
